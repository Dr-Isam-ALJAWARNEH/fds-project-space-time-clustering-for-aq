```python
def categorize_pm25(pm):
    if pm <= 12.0:
        return 'Low'
    elif pm <= 35.4:
        return 'Medium'
    else:
        return 'High'
```

```python
# Apply to table
joined_sorted['pollution_category'] = joined_sorted['PM25'].apply(categorize_pm25)

# Preview
joined_sorted[['PM25', 'pollution_category']].head()
```

```python
# Show the full table with the new pollution category column
joined_sorted[['geohash', 'time_bin', 'pickup_trip_count', 'dropoff_trip_count', 'total_trip_count',
               'Temperature', 'Humidity', 'PM25', 'pollution_category', 'CO', 'latitude', 'longitude',
               'geometry', 'index_right', 'community']].head(20)
```

```python
import folium
from folium.plugins import MarkerCluster
```

```python
# Center map over Chicago
chicago_center = [41.8781, -87.6298]
m = folium.Map(location=chicago_center, zoom_start=11)
```

```python
# Set colors for different pollution levels
color_dict = {
    'Low': 'green',
    'Medium': 'orange',
    'High': 'red'
}
```

```python
# Create a cluster layer
marker_cluster = MarkerCluster().add_to(m)

# Loop through your FULL joined_sorted data
for idx, row in joined_sorted.iterrows():
    folium.CircleMarker(
        location=[row['latitude'], row['longitude']],
        radius=4,
        color=color_dict.get(row['pollution_category'], 'blue'),
        fill=True,
        fill_color=color_dict.get(row['pollution_category'], 'blue'),
        fill_opacity=0.7,
        popup=(f"PM2.5: {row['PM25']:.2f}\n"
               f"Trips: {row['total_trip_count']}")
    ).add_to(marker_cluster)

# Display the map
m
```
